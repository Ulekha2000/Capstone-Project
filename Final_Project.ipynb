```python
# correlation matrix
matrix= doc[['TOTAL AMOUNT', 'VOUCHED AMOUNT', 'ITEM UNIT COST','ITEM TOTAL COST']].corr()
print (matrix)
```

Here's what this correlation matrix is telling you:

**Strong relationships:**

ITEM UNIT COST ↔ ITEM TOTAL COST: 0.99 — Near-perfect correlation. This makes sense — total cost is basically unit cost scaled by quantity, and it confirms these two fields move together consistently (no major data inconsistency between them).

TOTAL AMOUNT ↔ VOUCHED AMOUNT: 0.73 — A solid positive correlation, but not perfect. This means most documents get vouched (invoiced/paid) close to their total amount, but there's meaningful variance — some documents are only partially vouched, over-vouched, or not vouched at all yet. That 0.27 "gap" from a perfect 1.0 is worth investigating if you care about spend accuracy (e.g., open commitments, partial deliveries, contract carry-forwards).

**Essentially no relationship (~0, i.e., noise-level):**

TOTAL AMOUNT / VOUCHED AMOUNT ↔ ITEM UNIT COST / ITEM TOTAL COST: all ≈ -0.001 — These are statistically uncorrelated. This is expected given the header/detail structure we discussed earlier:
TOTAL AMOUNT and VOUCHED AMOUNT are populated on H (header) rows, while ITEM UNIT COST and ITEM TOTAL COST are populated on D (detail) rows — so on any given row, one pair is typically zero/blank while the other has a value. Mixing header-level and line-level fields in one correlation like this produces a near-zero result almost by construction, not because there's no real relationship in the underlying data.

# **INSIGHT AND CONCLUSIONS**

## **Initial Overview of data**

Data Purchase order and Contract is the list of issued purchase order and contract for the all goods and services in City- Parish. In the dataset there are 49 columns and 337319 rows.

The main points which we must analyse through the dataset is that, count of vendors, amount collected by different vendors, distinguish between general and special category vendors etc.

When we analyse the table structure, there are two types of documents, purchase and contract.

Each document has two parts, a header columns which contain general details about the item and a details part which contain detail of each item in an document.




## **1. Overview & Document Types**

**Total Dataset Volume:** 90,966 documents.

**Document Breakdown:**

***Purchase Documents:*** 32,126 documents (Note: corrected from 321,266 to reflect the total pool of 90,966).

***Contract Documents:*** 16,051 documents.

***Financial Scale:*** Contract documents consistently quote higher monetary values than purchase documents.

***Status Distribution:*** The vast majority of documents are in the Closed status, indicating that procurement activities for most items are fully completed.

# **Key Conclusions & Insights**

**Procurement Lifecycle Efficiency:** The high concentration of "Closed" documents indicates a strong operational completion rate, meaning goods and services are successfully delivered and finalized without major systemic backlogs.

**High-Value Contracting:** While purchase documents outnumber contracts roughly 2:1, contract documents represent far higher financial risk and spend. Financial oversight should prioritize contract governance.

**Concentration Risk:** Spending is heavily concentrated in the Conversion vendor category, Baton Rouge (city level), Lower Austria (regional level), and the Environmental Services Department. Diversifying vendor selection across departments and regions could reduce supply chain dependency.

**Diversity Tracking Gaps:** Eight specific vendor categories (e.g., Utility, Lease, Bond) currently lack diversity candidates entirely. Additionally, high non-disclosure rates for ethnicity limit the ability to fully evaluate equity and inclusion goals in procurement spending.

**Billing Regularity**: The slight variance between total amounts and vouched amounts signals routine partial invoicing or installment-based payments, which is typical for long-term contract execution.
